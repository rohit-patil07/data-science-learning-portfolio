1. Difference between a Function and a Method in Python

**Function :**

A Function is a  block of code that performs a specific task.

Function Can be called independently.

**Method :**

A Method is a function that belongs to an object/class.

Method can be Called using an object.

```python
#EXAMPLE
def add(a, b):        # function
    return a + b

s = "hello"
s.upper()             # upper() is a method
```

2. Function Arguments and Parameters

**Parameters :**

Parameters are the variables written in function definition.

**Arguments :**

Arguments are the actual values passed when calling the function.


```python
#EXAMPLE
def greet(name):   # name = parameter
    print("Hello", name)

greet("Rohit")     # "Rohit" = argument
```

3. Different Ways to Define & Call a Function

**Normal Function :**

A normal function is defined using the def keyword and performs a specific task.
It helps organize code and makes the program easier to understand and manage.


**Function with Parameters :**

A function with parameters accepts input values while it is called.
This allows the same function to work with different data and makes it more flexible.

**Lambda Function :**

A lambda function is a small, anonymous function written in a single line.
It is mainly used for simple and short operations where a full function definition is not required.

**Function with Default Argument :**

A function with a default argument has predefined values for its parameters.
If no value is passed during the function call, the default value is used automatically.

```python
#Normal function

def fun():
    print("Hello")
fun()


#Function with parameters

def add(a, b):
    return a + b
add(5, 6)


#Lambda function

square = lambda x: x*x
square(4)


#Function with default argument

def greet(name="User"):
    print("Hello", name)
greet()
```

4. Purpose of return Statement

Return sends back a value from the function.

It helps reuse the output of a function in other parts of the program

It makes functions more useful for calculations and data processing

Return statement is used to End the function execution.



```python
#EXAMPLE
def multiply(a, b):
    return a * b
print(multiply(4, 5))
```

5. Iterators vs Iterables

**Iterable:**

An iterable is an object that can be looped over

It contains multiple values

It can be used in a for loop

It does not store the current position of iteration

Examples include list, tuple, set, and string

**Iterator:**

An iterator is an object used to iterate over an iterable

It keeps track of the current position during iteration

It returns values one at a time

Example:


```python
nums = [1,2,3]       # iterable
it = iter(nums)      # iterator
print(next(it))
```

6. Generators in Python

Generators are functions that give values one at a time using yield.

They don’t store all values in memory, so they save space.

Use yield instead of return to pause and resume the function.

Values are created only when needed (lazy evaluation).

Good for large data or infinite sequences, like reading big files line by line.

Example:

```python
def counter():
    for i in range(8):
        yield i
```

7. Advantages of Generators

Memory efficient

Faster for large data

Generates values on demand

Useful for infinite sequences

8. Lambda Function

Small anonymous function.

Used for short, simple operations.

Used with map(), filter(), reduce().

Good for quick, simple operations without writing a full function.

EXAMPLE:

```python
add = lambda a, b: a + b
print(add(2, 3))
```

9. Purpose of map() Function

map() is used to apply a function to every item of a list (or any iterable) automatically.

It returns a map object, which can be converted to a list or tuple.

Useful to perform the same operation on multiple items without writing a loop.

Example: Convert numbers to squares

```python
nums = [1,2,3]
result = list(map(lambda x: x*x, nums))
```

10. Difference Between map(), filter(), reduce()

**Function Purpose:**

map()	    Applies function to all items

filter()	Selects items based on condition

reduce()	Reduces a list to a single value

Example:

```python
from functools import reduce

map(lambda x: x*2, [1,2,3])
filter(lambda x: x>2, [1,2,3])
reduce(lambda a,b: a+b, [1,2,3])
```

```python
#1. Function to return sum of even numbers
def sum_even(nums):
    return sum([n for n in nums if n % 2 == 0])

print(sum_even([1,2,3,4,5,6]))

#2. Reverse a String
def reverse_string(s):
    return s[::-1]

print(reverse_string("Python"))

#3. Square of Each Number in a List
def squares(nums):
    return [n*n for n in nums]

print(squares([1,2,3]))

#4. Check Prime Numbers from 1 to 200
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

primes = [n for n in range(1,201) if is_prime(n)]
print(primes)

#5. Iterator class for Fibonacci Sequence
class Fibonacci:
    def __init__(self, terms):
        self.terms = terms
        self.a = 0
        self.b = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.terms:
            raise StopIteration
        value = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return value

for x in Fibonacci(10):
    print(x)

#6. Generator for Powers of 2
def power_of_two(n):
    for i in range(n+1):
        yield 2 ** i

for num in power_of_two(5):
    print(num)

#7. Generator to Read File Line by Line
def read_file(filename):
    with open(filename, "r") as f:
        for line in f:
            yield line.strip()

#8. Sort List of Tuples by Second Value
data = [(1,4), (3,1), (2,7)]
sorted_data = sorted(data, key=lambda x: x[1])
print(sorted_data)

#9. Convert Celsius to Fahrenheit Using map()
celsius = [0,10,25,40]
fahrenheit = list(map(lambda c: (c*9/5)+32, celsius))
print(fahrenheit)

#10. Remove Vowels Using filter()
def remove_vowels(s):
    vowels = "aeiouAEIOU"
    return ''.join(filter(lambda ch: ch not in vowels, s))

print(remove_vowels("Rohit Patil"))

#11. Book Shop Accounting Using map() + lambda

#List example: [ [34587, 4, 40.95], [98762, 5, 10.95], ...]
#where: order_number, quantity, price_per_item

orders = [
    [34587, 4, 40.95],
    [98762, 5, 10.95],
    [77226, 3, 55.50],
    [88112, 2, 18.00]
]

result = list(map(lambda x:
                  (x[0], x[1]*x[2] if x[1]*x[2] >= 100 else x[1]*x[2] + 10),
                  orders))

print(result)
```

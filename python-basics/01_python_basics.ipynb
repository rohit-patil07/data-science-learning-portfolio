## Mutable vs Immutable
Lists are mutable; tuples are immutable.

```
lst = [1, 2, 3]
lst[0] = 99
lst
```

```
tpl = (1, 2, 3)
# tpl[0] = 99  # This would raise an error

tpl
```

## Conditional Statements
Using `if-elif-else` with sample values.

```
x = 7
if x > 10:
    result = 'Greater than 10'
elif x == 10:
    result = 'Equal to 10'
else:
    result = 'Less than 10'
result
```

2. How does broadcasting work in NumPy?

-> Broadcasting in NumPy allows operations between arrays of different shapes. It automatically expands the smaller array to match the size of the larger array so calculations can be performed easily.

It helps to perform operations without using loops and makes computation faster.

Rules of Broadcasting:

NumPy compares array shapes from right to left.

Dimensions must be equal or one of them must be 1.

If compatible, the smaller array is automatically stretched.

17. Difference between apply() and map() in Pandas

-> -apply()

Used on DataFrame and Series.

Applies function to rows or columns.

More flexible and powerful.

Can perform complex operations.

Works row-wise (axis=1) or column-wise (axis=0).

-map()

Used only on Series (single column).

Applies function to each element.

Used for simple transformations.

Less flexible compared to apply().

Performs element-wise operation.

21. Why NumPy slicing is faster than list slicing

-> NumPy slicing is faster than Python list slicing because NumPy arrays store data in continuous memory locations and use optimized internal operations, while Python lists require more processing.

Reasons why NumPy slicing is faster:

Continuous Memory Storage → NumPy stores data in contiguous memory, so access is faster.

No Copying of Data → NumPy slicing returns a view (reference) of the original array.

Optimized Implementation → Uses efficient internal C-based operations.

Less Memory Usage → More efficient memory handling.

Faster Computation → Designed for high-performance numerical operations.

```python
#4. Correlation matrix using Seaborn and visualize with heatmap
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# sample DataFrame
df = pd.DataFrame(np.random.randn(100, 4), columns=list('ABCD'))

corr = df.corr()
print("Correlation matrix:\n", corr)

sns.heatmap(corr, annot=True, cmap='viridis')
plt.title("Correlation Heatmap")
plt.show()
```

## What are Data Structures?
Data structures are ways to store and organize data so that it can be accessed and modified easily.
In data science, data structures are heavily used before converting data into DataFrames.

Common Python data structures:

>List

>Tuple

>Set

>Dictionary

##Lists

Lists are ordered, mutable collections that can store multiple values.

```python
#Creating an list named numbers
numbers = [10, 20, 30, 40]
numbers

#Accessing Elements
numbers[0]
```

```python
#Modifying Elements
numbers[1] = 25
numbers
```

**Lists are commonly used to store:**

Numeric values

Dataset columns

Intermediate results

##Tuples
Tuples are ordered but immutable collections.
Once created, their values cannot be changed.

```python
coordinates = (10, 20)
coordinates
```

**Tuples are useful when:**

Data should not be modified

Fixed values like coordinates or configuration data are required

##Sets
Sets store unique values only and are unordered.

```python
unique_values = {1, 2, 3, 3, 4}
unique_values
```

**Sets are useful for:**

Removing duplicate values

Membership testing

Mathematical set operations

##Dictionaries
Dictionaries store data in key–value pairs.
They are very important in data science for structured data.

```python
student = {
    "name": "Rohit",
    "age": 22,
    "course": "Data Science"
}
student
```

```python
#Accessing Values
student["name"]
```

```python
#Updating Value
student["age"] = 23
student
```

**Dictionaries are often used for:**

JSON-like data

Storing records

API responses

**Looping Through Data Structures :**

```python
#Loop through a List
for num in numbers:
    print(num)
```

```python
#Loop through a Dictionary
for key, value in student.items():
    print(key, ":", value)
```

Loops help process large datasets efficiently

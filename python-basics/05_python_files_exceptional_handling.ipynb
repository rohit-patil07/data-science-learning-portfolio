1. What is the difference between interpreted and compiled languages?

-> Compiled language:

Code is converted into machine code before execution.

Faster execution.

Example: C, C++.

Interpreted language:

Code is translated line-by-line at runtime.

Slower execution.

Example: Python, JavaScript.

2. What is exception handling in Python?

-> It’s a way to deal with errors in your program without crashing it.

You “try” a block of code, and if an error occurs, it goes to the “except” block where you handle it.

Helps your program continue running even if something goes wrong.

3. Purpose of the finally block?

-> finally block always runs no matter what, whether an exception occurred or not.

Useful for cleanup tasks like closing files, releasing resources, etc.

4. What is logging in Python?

-> Logging is a way to record events, errors, and debug information in your program.

It can save messages to a file or show them on the console. Using different log levels like DEBUG or ERROR helps track issues easily.

5. Significance of the __del__ method

-> __del__() is a destructor called when an object is destroyed. It helps clean up resources like files or database connections. This prevents memory leaks and keeps the program safe.

6. Difference between import and from ... import

-> import module loads the whole module, so you use it like module.func().

from module import func brings in only a specific function, letting you use it directly. This keeps code cleaner and lighter.

7. How to handle multiple exceptions in Python

-> You can use multiple except blocks to handle different errors separately.

Each block deals with a specific type, like ValueError or TypeError. This keeps the program from crashing and handles errors gracefully.

8. Purpose of with statement in file handling

-> with opens a file and automatically closes it when done.

This works even if an error occurs while working with the file. It makes file handling safe and avoids data loss.

9. Difference between multithreading and multiprocessing

-> Multithreading runs multiple threads in one process and shares memory, ideal for I/O tasks.

Multiprocessing runs separate processes with independent memory, best for CPU-heavy tasks. Threads are lightweight, processes are more powerful.

10. Advantages of logging

-> Logging helps track program events and errors without stopping the program.

It keeps a record for debugging and organizes messages using levels like INFO or WARNING. This makes maintenance much easier.

11. What is memory management in Python?

-> Python automatically manages memory using a private heap and garbage collector. Objects are deleted when no longer used. Reference counting helps track memory efficiently.

12. Basic steps in exception handling

-> You put risky code inside a try block. Errors are caught in except, while else runs if there’s no error. finally always runs for cleanup.

13. Why is memory management important?

-> It prevents memory leaks and keeps programs fast. Proper management avoids crashes and ensures resources like files are released.

14. Role of try and except

-> try contains code that might fail. except handles the error so the program doesn’t crash. Together, they make the program robust.

15. How Python’s garbage collection works

-> Python uses reference counting to track objects. The garbage collector removes unused objects. This keeps memory free and avoids leaks.

16. Purpose of else block in exception handling

-> else runs only if no exception occurs in try. It separates normal code flow from error handling. Useful for clarity and organization.

17. Common logging levels

-> DEBUG for detailed info, INFO for general messages. WARNING for potential issues, ERROR for serious problems, and CRITICAL for very severe errors.

18. Difference between os.fork() and multiprocessing

-> os.fork() duplicates a process on Unix only. Multiprocessing works cross-platform with separate memory. Multiprocessing is easier and safer for parallel tasks.

19. Importance of closing a file

-> Closing a file frees system resources and ensures data is saved properly. It prevents corruption and keeps programs stable.

20. Difference between file.read() and file.readline()

-> read() reads the whole file at once. readline() reads one line at a time. Use readline() for big files to save memory.

21. Purpose of logging module

-> Used to record information, errors, warnings, and debug messages. Helps track program events and makes debugging easier.

22. Purpose of os module in file handling

-> Helps work with file paths, create or delete directories, and check if files exist. Makes file management in Python easier.

23. Challenges in memory management

-> Circular references and memory leaks can cause problems. Large objects and unused variables need careful handling to avoid slowdowns or crashes.

24. How to raise an exception manually?

-> Use raise to throw an error yourself. Example: raise ValueError("Invalid value"). Useful to enforce rules or check input.

25. Why is multithreading important?

-> Runs multiple tasks at the same time, improving performance. Useful for I/O tasks like downloads or file reading, keeps programs responsive.

```python
# Q18. How would you implement a basic logging setup that logs to a file with rotation after 1MB?
import logging
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler("rotating.log", maxBytes=1_000_000, backupCount=3)
logger = logging.getLogger()
logger.setLevel(logging.INFO)
logger.addHandler(handler)

logger.info("This is a log message")
```

```python
# Q19. Write a program that handles both IndexError and KeyError using a try-except block.
my_list = [1, 2, 3]
my_dict = {"name": "Rohit"}

try:
    print(my_list[5])
    print(my_dict["age"])
except IndexError:
    print("Index out of range!")
except KeyError:
    print("Key does not exist!")  # Handles both types of errors
```

```python
# Q21. Write a Python program that reads a file and prints the number of occurrences of a specific word.
word_count = 0
target_word = "Python"
with open("example.txt", "r") as file:
    for line in file:
        word_count += line.lower().count(target_word.lower())
print(f"'{target_word}' occurs {word_count} times")
```
